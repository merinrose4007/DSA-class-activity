Load into SQLite:

```python
#● Create an SQLite database
import sqlite3
conn = sqlite3.connect("country.db")
cursor = conn.cursor()
```

```python
#● Load the final merged dataset into a database table
data.to_sql("final_dataset", conn, if_exists="replace", index=False)
```

```python
result = pd.read_sql("SELECT * FROM final_dataset", conn)
print(result.head())
```

```python
#Countries with highest internet penetration
cursor.execute("SELECT country FROM final_dataset order by internet_penetration_rate desc limit 1")
print(cursor.fetchall())
```

```python
# Average literacy rate
cursor.execute("select avg(literacy_rate) from final_dataset")
print(cursor.fetchall())
```

```python
# Correlation between literacy rate and internet penetration

cursor.execute("SELECT (COUNT(*) * SUM(literacy_rate * internet_penetration_rate) - SUM(literacy_rate) * SUM(internet_penetration_rate))/ SQRT((COUNT(*) * SUM(literacy_rate * literacy_rate) - POWER(SUM(literacy_rate), 2)) *(COUNT(*) * SUM(internet_penetration_rate * internet_penetration_rate) - POWER(SUM(internet_penetration_rate), 2))) AS correlation FROM final_dataset;")
print(cursor.fetchall())
```

```python
#GDP per capita analysis
cursor.execute("select country, GDP/population as gdp_per_capita from final_dataset ORDER BY gdp_per_capita DESC")
print(cursor.fetchall())
```

```python
# Which countries have a GDP per capita above $10,000?
cursor.execute("select country, GDP/population as gdp_per_capita from final_dataset where gdp_per_capita>10000")
print(cursor.fetchall())
```

```python
# What is the total population covered in the dataset?
cursor.execute("select sum(population) from final_dataset")
print(cursor.fetchall())
```

```python
#Which countries have the lowest literacy rates, and how does that impact internet access?

cursor.execute("select country, literacy_rate, internet_penetration_rate from final_dataset order by literacy_rate asc limit 1")
print(cursor.fetchall())

# Saudi Arabia has the lowest literacy rate hence their internet usage is also low
```

```python
#What are the top 5 wealthiest countries by total GDP, and how does that compare with population size?

cursor.execute("select country, GDP, population, GDP/population as gdp_per_capita from final_dataset order by GDP desc limit 5")
print(cursor.fetchall())

# For India and Japan Huge populations dilute per-person wealth.
# For Italy and Spain Smaller populations mean that even moderately high GDP translates into higher wealth per person.
```

```python
# Find countries where internet users exceed 70% of the population
cursor.execute("""SELECT
    country,
    (internet_users * 100.0 / population) AS percent
FROM final_dataset
WHERE percent > 70""")
print(cursor.fetchall())
```

```python
# What is the average GDP per capita for countries with internet penetration above 50%?

cursor.execute("select avg(GDP/population) as gdp_per_capita from final_dataset where internet_penetration_rate>50")
print(cursor.fetchall())
```

```python
# How many countries have a literacy rate above 90%, and what is their average internet penetration rate?
cursor.execute('select count(*), avg(internet_penetration_rate) from final_dataset where literacy_rate>90')
print(cursor.fetchall())
```
