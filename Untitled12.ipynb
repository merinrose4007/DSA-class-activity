Data Cleaning

Since Review Scores Rating and Review Scores Rating bin are both skewed the missing values were filled using median. Since zipcode can be similar for similar neighbourhood they were filled by goruping the neighbourhood and filling them using mode. Other features were also filled using mode


```python
data['Number Of Reviews'].value_counts()
```

```python
data['Review Scores Rating'].unique()
```

```python
data['Review Scores Rating'].hist()
```

```python
data['Review Scores Rating (bin)'].hist()
```

```python
data['Review Scores Rating'] = data['Review Scores Rating'].fillna(data['Review Scores Rating'].median())
```

```python
data['Review Scores Rating (bin)'] = data['Review Scores Rating (bin)'].fillna(data['Review Scores Rating (bin)'].median())
```

```python
data.isna().sum()
```

```python
data['Neighbourhood '].unique()
```

```python
data[data['Neighbourhood ']=='Manhattan']
```

```python
for i in data['Neighbourhood '].unique():
  data.loc[data['Neighbourhood ']==i, 'Zipcode'] = data.loc[data['Neighbourhood ']==i, 'Zipcode'].fillna(data.loc[data['Neighbourhood ']==i, 'Zipcode'].mode()[0])
```

```python
data.isna().sum()
```

```python
data['Property Type'] = data['Property Type'].fillna(data['Property Type'].mode()[0])
```

```python
data['Beds'].unique()
```

```python
data[data['Beds'] == 3]
```

```python
data['Beds'] = data['Beds'].fillna(data['Beds'].mode()[0])
```
